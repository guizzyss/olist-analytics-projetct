```python
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv

# 1. Carrega as variáveis de ambiente (as mesmas usadas no load.py)
load_dotenv()

DB_USER = os.getenv("DB_USER")
DB_PASS = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST", "localhost")
DB_PORT = os.getenv("DB_PORT", "5433") # Lembre-se que configuramos o Docker na 5433!
DB_NAME = os.getenv("DB_NAME", "olist_db")

# 2. Cria o motor de conexão com o PostgreSQL
string_conexao = f"postgresql+psycopg://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = create_engine(string_conexao)

print("✅ Conectado ao banco de dados local com sucesso!")
```

```python
#Mapeia os pedidos que têm mais de 1 linha de pagamento registrada
query_multiplos_pagamentos = """
SELECT 
        order_id, 
        COUNT(*) AS qtd_transacoes, 
        SUM(payment_value) AS valor_total_pedido
    FROM fato_pagamentos
    GROUP BY order_id
    HAVING COUNT(*) > 1
    ORDER BY qtd_transacoes DESC
    LIMIT 5;
    """

df_multiplos = pd.read_sql(query_multiplos_pagamentos, con=engine)

print("✅ Consulta executada com sucesso! Aqui estão os pedidos com múltiplos pagamentos:")
print(df_multiplos)
```

```python
id_produto_investigado = "fa65dad1b0e818e3ccc5cb0e39231352"

query_detalhe_pagamento = f"""
    SELECT 
        order_id, 
        payment_sequential, 
        payment_type, 
        payment_installments, 
        payment_value
    FROM fato_pagamentos
    WHERE order_id = '{id_produto_investigado}'
    ORDER BY payment_sequential;
"""

df_detalhe = pd.read_sql(query_detalhe_pagamento, con=engine)

print(f"🔍 DETALHES DOS PAGAMENTOS DO PEDIDO: {id_produto_investigado}")
display(df_detalhe)
```
