```python
import os
import pandas as pd
import kagglehub

# 1. O kagglehub identifica que você já baixou e apenas retorna o caminho exato!
caminho_dados = kagglehub.dataset_download("olistbr/brazilian-ecommerce")
print(f"📂 Os dados estão escondidos nesta pasta do seu PC:\n{caminho_dados}\n")

# 2. Agora pedimos para o Python listar tudo o que termina com .csv dentro dessa pasta
arquivos = [f for f in os.listdir(caminho_dados) if f.endswith('.csv')]

print("📊 Arquivos CSV encontrados no Dataset da Olist:")
for arquivo in arquivos:
    print(f" - {arquivo}")
```

```python
df_pedidos = pd.read_csv(os.path.join(caminho_dados, "olist_orders_dataset.csv"))
df_itens = pd.read_csv(os.path.join(caminho_dados, "olist_order_items_dataset.csv"))

print("Tabelas carregadas com sucesso!")
```
