```python
import os
import pandas as pd
import kagglehub

caminho_dados = kagglehub.dataset_download("olistbr/brazilian-ecommerce")
print(f"Diretório base: {caminho_dados}\n")

arquivos = [f for f in os.listdir(caminho_dados) if f.endswith('.csv')]
print(f"Arquivos CSV disponíveis no dataset:")
for arquivo in arquivos:
    print(f" - {arquivo}")
```

```python
df_pedidos = pd.read_csv(os.path.join(caminho_dados, "olist_orders_dataset.csv"))
df_itens = pd.read_csv(os.path.join(caminho_dados, "olist_order_items_dataset.csv"))

print("DataFrames 'df_pedidos' e 'df_itens' carregados na memória!")
```

```python
print("--- AMOSTRA: TABELA DE PEDIDOS ---")
display(df_pedidos.head(3))

print("\n--- ESTRUTURA DOS DADOS ---")
df_pedidos.info()

print("\n--- CONTAGEM DE VALORES NULOS ---")
display(df_pedidos.isnull().sum().to_frame("Nulos").T)
```

```python
print("--- AMOSTRA: TABELA DE ITENS ---")
display(df_itens.head(3))

print("\n--- ESTRUTURA DOS DADOS ---")
df_itens.info()

print("\n--- CONTAGEM DE VALORES NULOS ---")
display(df_itens.isnull().sum().to_frame("Nulos").T)
```
